# Book Recommendation System - Data Exploration

This notebook explores the book dataset and prepares it for recommendation algorithms.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
```

```python
# Load sample data (you'll need to replace with actual dataset)
# Popular datasets: Goodreads, Amazon Books, Book-Crossing

# Create sample data for demonstration
books_data = {
    'book_id': range(1, 101),
    'title': [f'Book {i}' for i in range(1, 101)],
    'author': [f'Author {i%20}' for i in range(1, 101)],
    'genre': np.random.choice(['Fiction', 'Non-Fiction', 'Mystery', 'Romance', 'Sci-Fi'], 100),
    'rating': np.random.uniform(3.0, 5.0, 100),
    'description': [f'Description for book {i}' for i in range(1, 101)]
}

books_df = pd.DataFrame(books_data)
print("Books Dataset Shape:", books_df.shape)
books_df.head()
```

```python
# Data exploration
print("Dataset Info:")
print(books_df.info())
print("\nMissing Values:")
print(books_df.isnull().sum())
```

```python
# Visualizations
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Rating distribution
axes[0,0].hist(books_df['rating'], bins=20, alpha=0.7)
axes[0,0].set_title('Rating Distribution')
axes[0,0].set_xlabel('Rating')

# Genre distribution
books_df['genre'].value_counts().plot(kind='bar', ax=axes[0,1])
axes[0,1].set_title('Genre Distribution')
axes[0,1].tick_params(axis='x', rotation=45)

# Top authors
books_df['author'].value_counts().head(10).plot(kind='bar', ax=axes[1,0])
axes[1,0].set_title('Top 10 Authors')
axes[1,0].tick_params(axis='x', rotation=45)

# Rating by genre
sns.boxplot(data=books_df, x='genre', y='rating', ax=axes[1,1])
axes[1,1].set_title('Rating by Genre')
axes[1,1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
